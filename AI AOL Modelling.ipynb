```python
# import libraries
from ultralytics import YOLO

dataSource = '/content/drive/MyDrive/AI AOL/data.yaml'
```

```python
# Load pretrained YOLOv8 model
model = YOLO("yolov8s.pt")
```

```python
# training
model.train(
    data=dataSource,
    epochs=100,
    imgsz=640,
    batch=16,
    name="train10_5000v7"
)
```

```python
# Evaluation
model.val(
    data=dataSource,
    imgsz=640
)
```

```python
# Inference / Prediction
model = YOLO("/content/drive/MyDrive/AI AOL/runs/detect/train10_5000v7/weights/best.pt")

model.predict(
    source="/content/drive/MyDrive/AI AOL/dataset/test/images/-1-2_jpg.rf.8abf9a5a00e0df1f26a57fccdc944704.jpg",
    imgsz=640,
    save=True
)
```
